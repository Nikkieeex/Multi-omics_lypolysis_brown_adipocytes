```python
# De manier van inlezen is gebasseerd op hoe Nicole en Gea het in hun Rmd logboeken hebben gedaan.
# Zie het logboek van Nicole voor het originele concept van data inlezen van proteomics en die van Gea voor transcriptomics


prot_df = pd.read_csv(
    "/students/2025-2026/Thema08/lipolyse/unpacked/proteomics/proteinGroups.txt",
    sep="\t",
    low_memory=False
)

# Selecteer alleen de LFQ intensity kolommen
lfq_cols = [c for c in prot_df.columns if c.startswith("LFQ intensity")]
lfq_df = prot_df[lfq_cols].copy()

# Nul-waarden -> NaN
lfq_df.replace(0, np.nan, inplace=True)

P = lfq_df.to_numpy(dtype=np.float64)
P_genes = prot_df["Gene names"].to_numpy()
P_samples = np.array(lfq_cols)

print("Proteomics shape:", P.shape)
# Verwacht: (20015, 18)  — eiwitten x samples

counts_df = pd.read_csv(
    "/students/2025-2026/Thema08/lipolyse/unpacked/transcriptomics/counts.txt",
    sep="\t",
    comment="#",      # slaat regels over die beginnen met #
    index_col=0       # GeneID als index
)

# Eerste 5 kolommen zijn metadata (Chr, Start, End, Strand, Length)
# Kolommen 5 t/m einde zijn de count-waarden per sample
G = counts_df.iloc[:, 5:].to_numpy(dtype=np.float64)
G_genes = counts_df.index.to_numpy()
G_samples = counts_df.columns[5:].to_numpy()

print("Transcriptomics shape:", G.shape)
# Verwacht: (57186, 18)  — genen x samples
```
